```python
import pandas as pd

df = pd.read_csv('btc_usdt_1d_full.csv')

# data.index as DatetimeIndex
df.set_index('timestamp', inplace=True)
df.index = pd.to_datetime(df.index) 

# visualize as candles using mplfinance
import mplfinance as mpf

mpf.plot(df, type='candle', style='charles', volume=True, title='BTC/USDT Daily Candles', ylabel='Price (USDT)', ylabel_lower='Volume', figratio=(12,8), figscale=1.5)
```
